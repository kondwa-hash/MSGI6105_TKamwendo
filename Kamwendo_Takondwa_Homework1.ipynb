# Calculations

```python
# Rectangle dimensions in metres
length_m = 12.5
width_m = 8.0

# Calculating area
area_m2 = length_m * width_m

# result
print("Area =", area_m2, "square metres")
```

```python
# Changing the length
length_m = 15.0

# Recalculating area
area_m2 = length_m * width_m

# New result
print("New area =", area_m2, "square metres")
```

The result changed because after updating 'length_m' , I ran the calculation cell again, which replaced the old 'area_m2' value with the new one. Just changing a variable doesn't automatically update past calculations. Python runs code one cell at a time, so any calculation using that variable has to be run again to give a new answer. Until you do that, the old result stays the same.
